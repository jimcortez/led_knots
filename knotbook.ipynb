```python
%matplotlib ipympl

import pyknotid.spacecurves as sp
import pyknotid.make as mk
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from led_knots.core import (
    scale_pyknot_points,
)
```

```python
# k = sp.Knot(mk.trefoil(num_points=100))
k = mk.k6_1(num_points=150)
```

```python
points = k.points
points_scaled = scale_pyknot_points(
    k.points,
    width=200,
    height=200,
    length=200,
)
```

```python

fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.plot(points[:, 0], points[:, 1], points[:, 2])
fig.show()
```

```python
fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.plot(points_scaled[:, 0], points_scaled[:, 1], points_scaled[:, 2])
fig.show()
```
