```python
%matplotlib ipympl

import pyknotid.spacecurves as sp
from pyknotid.representations import DTNotation, PlanarDiagram, GaussCode
from pyknotid.catalogue import identify, get_knot, from_invariants  
import pyknotid.make as mk
from pyknotid.make.periodic_knot import simple_link2
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from led_knots.core import (
    scale_pyknot_points, dowker_to_knot, relax_knot_points
)
import numpy as np

# The 15 knots below are the set this project builds. Each cell names the
# module (src/led_knots/knots/) and config (knot_configs/) that render it:
#     uv run render-knot knot_configs/<config>.yaml
# Tweak an individual knot by editing its config, not this notebook.

num_points = 200

# Slots 12-15 need a finer curve. relax_knot_points moves strands past each
# other when the curve is coarser than the gaps between strands, and above 11
# crossings 200 points is coarse enough that the relaxed curve comes out as a
# different, much simpler knot -- silently. At 200, K15n3 identifies as 6_2.
dense_points = 600

scale = (200, 200, 200)
```

```python
def show_points(points, title="", scale=None):
    if scale is not None:
        points = scale_pyknot_points(
            points,
            width=scale[0],
            height=scale[1],
            length=scale[2],
        )
    points = np.asarray(points)
    fig = plt.figure()
    ax = fig.add_subplot(projection='3d')
    ax.plot(points[:, 0], points[:, 1], points[:, 2])
    ax.set_title(title)
    
    plt.show()

    # ax.view_init(azim=225)

    # plt.show()
```

```python
# Knot 1: Ring (the unknot) -- knots/ring.py -- knot_configs/k1_1-ring.yaml
data = np.zeros((num_points, 3), dtype=np.float64)
ts = np.linspace(0, 2 * np.pi, num_points)
data[:, 0] = 3 * np.sin(ts)
data[:, 1] = 3 * np.cos(ts)
k = sp.Knot(data)
show_points(k.points, title="Ring Knot (k1_1)", scale=scale)
```

```python
# Knot 2: (2,1) torus knot -- knots/k2_1.py -- knot_configs/k2_1.yaml
# There is no 2-crossing knot; (2,1) is topologically the unknot with a twist
# in it. It fills this slot as the visual step from the flat ring to the trefoil.
k = mk.torus_knot(p=2, q=1, num=num_points)
show_points(k.points, title="k2_1", scale=scale)
```

```python
# Knot 3: Trefoil (3_1) -- knots/trefoil.py -- knot_configs/k3_1-trefoil.yaml
from pyknotid.make import trefoil as make_trefoil
k = make_trefoil(num_points=num_points)
show_points(k.points, title="Trefoil Knot (k3_1)", scale=scale)
```

```python
# Search for knots with a specific number of crossings.
# Scratch cell for picking the next slot; not one of the 15.
crossings= 15
#for i, inv in enumerate(from_invariants(symmetry="fully amphicheiral")):
for i, inv in enumerate(from_invariants(max_crossings=crossings, min_crossings=crossings)):
    inv.pretty_print()
    if inv.dt_code and inv.dt_code.strip() != "":
        dt_code = inv.dt_code.split()
        k = dowker_to_knot(dt_code, num_points=dense_points, relax_steps=400)
        show_points(k.points, title=f"Knot {inv.identifier}", scale=scale)
        if i >= 10:
            break
    else:
        print(f"No DT code for knot {inv.identifier}.")
else:
    print("No knots found.")
```
